```dockerfile
# Dockerfile for ML development
FROM python:3.11-slim

WORKDIR /workspace

# Install system dependencies
RUN apt-get update && apt-get install -y \
    build-essential git curl && \
    rm -rf /var/lib/apt/lists/*

# Install Python packages
COPY requirements.txt .
RUN pip install --no-cache-dir -r requirements.txt

# Expose Jupyter port
EXPOSE 8888

CMD ["jupyter", "lab", "--ip=0.0.0.0", "--no-browser", "--allow-root"]
```

**Example 1.9.5**

````python
import sys
import importlib

packages = {
    'numpy':        '1.24',
    'pandas':       '2.0',
    'matplotlib':   '3.7',
    'seaborn':      '0.12',
    'sklearn':      '1.3',
    'xgboost':      '1.7',
    'lightgbm':     '4.0',
}

print(f'Python {sys.version}\n')
print(f'{"Package":<20} {"Installed":<15} {"Status"}')
print('-' * 50)

all_ok = True
for pkg, min_ver in packages.items():
    try:
        mod = importlib.import_module(pkg)
        version = getattr(mod, '__version__', 'unknown')
        ok = version >= min_ver
        status = '✓  OK' if ok else '⚠  OLD VERSION'
        if not ok: all_ok = False
    except ImportError:
        version = 'NOT INSTALLED'
        status = '✗  MISSING'
        all_ok = False
    print(f'{pkg:<20} {version:<15} {status}')

print()
print('All checks passed!' if all_ok else 'Some packages need attention.')
````

**Example 2.2.2**

```python
import numpy as np

x = np.array([3.0, 4.0, 0.0])   # e.g. [height_cm, weight_kg, age_norm]
y = np.array([1.0, 2.0, 3.0])

print('x + y =', x + y)                         # [4. 6. 3.]
print('3 * x =', 3 * x)                          # [9. 12. 0.]
print('Dot product:', np.dot(x, y))              # 3×1 + 4×2 + 0×3 = 11.0
print('L2 norm:', np.linalg.norm(x))             # √(9+16+0) = 5.0
print('L1 norm:', np.linalg.norm(x, ord=1))      # 3+4+0 = 7.0

cos_sim = np.dot(x,y) / (np.linalg.norm(x) * np.linalg.norm(y))
print(f'Cosine similarity: {cos_sim:.4f}')       # 0.6285

# Unit vector: direction only, magnitude = 1
x_unit = x / np.linalg.norm(x)                  # [0.6, 0.8, 0.0]
print(f'Unit vector: {x_unit}  |length={np.linalg.norm(x_unit):.1f}|')
```

**Example 2.2.3**

```python
# 4 samples × 3 features: [age, income_k, years_experience]
X = np.array([[25, 40,  2],
              [32, 65,  8],
              [28, 52,  4],
              [45, 90, 20]], dtype=float)

print('Shape:          ', X.shape)        # (4, 3)
print('Num samples:    ', X.shape[0])     # 4
print('Num features:   ', X.shape[1])     # 3
print('X[1, 2] =       ', X[1, 2])        # 8.0 (row 1, col 2)
print('Row 0 (sample): ', X[0])           # [25. 40.  2.]
print('Col 1 (income): ', X[:, 1])        # [40. 65. 52. 90.]
print('Column means:   ', X.mean(axis=0)) # mean per feature
print('Row means:      ', X.mean(axis=1)) # mean per sample
```

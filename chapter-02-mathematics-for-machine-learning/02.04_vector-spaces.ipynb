**Example 2.4.1**

```python
A_full   = np.array([[1,0],[0,1],[1,1]])     # rank 2 - full rank
A_defect = np.array([[1,2],[2,4],[3,6]])     # rank 1 - col2 = 2×col1

print('rank(A_full):  ', np.linalg.matrix_rank(A_full))    # 2
print('rank(A_defect):', np.linalg.matrix_rank(A_defect))  # 1

# Detecting near-collinearity via condition number
# High condition number → near rank-deficient → numerical instability
A_illcond = np.array([[1.0, 1.001],[1.0, 1.0]])  # very similar columns
print(f'Condition number: {np.linalg.cond(A_illcond):.0f}')  # very high
print('Rule of thumb: condition number > 1e6 → serious collinearity')
```

**Example 2.5.1**

```python
A = np.array([[4., 1.],[2., 3.]])
eigenvalues, eigenvectors = np.linalg.eig(A)

print('Eigenvalues:  ', eigenvalues)          # [5. 2.]
print('Eigenvectors (columns):'); print(np.round(eigenvectors, 4))

# Verify: Av = λv
v1, lam1 = eigenvectors[:,0], eigenvalues[0]
print(f'\nAv₁  = {np.round(A@v1, 4)}')
print(f'λ₁v₁ = {np.round(lam1*v1, 4)}')
print(f'Match: {np.allclose(A@v1, lam1*v1)}')

# PCA via eigendecomposition of the covariance matrix
np.random.seed(0)
data = np.random.multivariate_normal([0,0], [[4,3],[3,3]], size=300)
C    = np.cov(data.T)                                # 2×2 covariance matrix
evals, evecs = np.linalg.eigh(C)                    # eigh: symmetric matrix
idx = np.argsort(evals)[::-1]                        # sort descending
evals, evecs = evals[idx], evecs[:,idx]

print(f'\nCovariance matrix:\n{np.round(C,3)}')
print(f'Eigenvalues (variance per PC): {np.round(evals,3)}')
print(f'Variance explained: PC1={evals[0]/evals.sum():.1%}, PC2={evals[1]/evals.sum():.1%}')

# Project data onto first principal component
data_pc1 = data @ evecs[:,0]   # shape (300,) - 1D representation
print(f'Data reduced from {data.shape} to {data_pc1.shape}')
```

**Example 2.6.1**

```python
# Activation functions and their derivatives
def sigmoid(x):       return 1 / (1 + np.exp(-x))
def sigmoid_d(x):     s=sigmoid(x); return s*(1-s)
def tanh_d(x):        return 1 - np.tanh(x)**2
def relu(x):          return np.maximum(0, x)
def relu_d(x):        return (x > 0).astype(float)
def leaky_relu(x, a=0.01):   return np.where(x>0, x, a*x)
def leaky_relu_d(x, a=0.01): return np.where(x>0, 1.0, a)

x = np.array([-2., -1., 0., 1., 2.])
print(f'x:          {x}')
print(f'sigmoid:    {np.round(sigmoid(x),4)}')
print(f'sigmoid d:  {np.round(sigmoid_d(x),4)}')
print(f'ReLU:       {relu(x)}')
print(f'ReLU d:     {relu_d(x)}')
print(f'Leaky ReLU: {leaky_relu(x)}')

# Numerical verification of sigmoid derivative at x=0
h = 1e-7
numerical = (sigmoid(0+h) - sigmoid(0-h)) / (2*h)
analytic  = sigmoid_d(0.0)
print(f'\nSigmoid derivative at x=0: analytic={analytic}, numerical={numerical:.7f}')
```

**Example 2.6.2**

```python
def f(w1, w2):   return 3*w1**2 + 2*w1*w2 + w2**3

def partial(f_func, idx, params, h=1e-7):
    p_hi, p_lo = list(params), list(params)
    p_hi[idx] += h; p_lo[idx] -= h
    return (f_func(*p_hi) - f_func(*p_lo)) / (2*h)

w1, w2 = 1.0, 2.0
df_dw1 = partial(f, 0, [w1, w2])
df_dw2 = partial(f, 1, [w1, w2])
print(f'f(1,2)   = {f(w1,w2)}')          # 15
print(f'∂f/∂w₁  = {df_dw1:.4f}')        # 10.0
print(f'∂f/∂w₂  = {df_dw2:.4f}')        # 14.0
```

**Example 2.6.3**

```python
# Manual backpropagation - single neuron, single sample
def sigmoid(x):    return 1/(1+np.exp(-x))
def sig_d(x):      s=sigmoid(x); return s*(1-s)

# Values
x, w, b, y = 2.0, 0.5, -1.0, 1.0

# Forward pass
z = w*x + b           # 0.0
a = sigmoid(z)        # 0.5
L = (a - y)**2        # 0.25
print(f'Forward:  z={z}  a={a}  L={L}')

# Backward pass (chain rule)
dL_da = 2*(a-y)       # -1.0
da_dz = sig_d(z)      #  0.25
dz_dw = x             #  2.0
dz_db = 1.0
dL_dw = dL_da * da_dz * dz_dw   # -0.5
dL_db = dL_da * da_dz * dz_db   # -0.25
print(f'Gradients: dL/dw={dL_dw}  dL/db={dL_db}')

# Gradient descent update
lr = 0.1
print(f'Updated:   w={w - lr*dL_dw:.4f}  b={b - lr*dL_db:.4f}')

# PyTorch autograd matches exactly
import torch
wt = torch.tensor([w], requires_grad=True)
bt = torch.tensor([b], requires_grad=True)
xt = torch.tensor([x])
yt = torch.tensor([y])
Lt = (torch.sigmoid(wt*xt+bt) - yt)**2
Lt.backward()
print(f'PyTorch:   dL/dw={wt.grad.item()}  dL/db={bt.grad.item()}')
```

**Example 2.7.1**

```python
# Gradient of a scalar loss over 2D parameter space
# L(w1, w2) = (w1 - 3)² + 2(w2 + 1)²  - minimum at (3, -1)
def L(params):   w1,w2=params; return (w1-3)**2 + 2*(w2+1)**2

def gradient_numerical(f, params, h=1e-5):
    grad = np.zeros_like(params, dtype=float)
    for i in range(len(params)):
        p_hi, p_lo = params.copy(), params.copy()
        p_hi[i]+=h; p_lo[i]-=h
        grad[i] = (f(p_hi)-f(p_lo))/(2*h)
    return grad

params = np.array([0.0, 0.0])
print(f'L at (0,0):{L(params)}')
print(f'Gradient at (0,0):{gradient_numerical(L, params)}')
print(f'(analytic: [2(w1-3), 4(w2+1)] = [-6, 4])')

# Simple gradient descent to find minimum
params = np.array([0.0, 0.0]); lr = 0.1
for step in range(30):
    g = gradient_numerical(L, params)
    params -= lr * g
print(f'\nAfter 30 steps: params={np.round(params,4)}')
print(f'True minimum:   params=[3.0, -1.0]')
print(f'Loss at minimum: {L(params):.6f}')
```

**Example 2.7.2**

```python
import torch

# Autograd example: compute gradients for a small neural network
torch.manual_seed(42)

# Parameters (requires_grad=True tells PyTorch to track these)
W1 = torch.randn(4, 3, requires_grad=True)   # Layer 1 weights
b1 = torch.zeros(4, requires_grad=True)       # Layer 1 bias
W2 = torch.randn(2, 4, requires_grad=True)   # Layer 2 weights
b2 = torch.zeros(2, requires_grad=True)       # Layer 2 bias

# One sample, 3 features
x = torch.randn(3)
y = torch.tensor([1.0, 0.0])  # True label (one-hot)

# Forward pass (PyTorch records the computation graph)
h  = torch.relu(W1 @ x + b1)          # Hidden layer
out = W2 @ h + b2                       # Output layer (logits)
loss = ((out - y)**2).sum()             # MSE loss

print(f'Loss: {loss.item():.4f}')

# Backward pass: one call computes ALL gradients
loss.backward()

print(f'W1.grad shape: {W1.grad.shape}')  # (4, 3) - gradient for each weight
print(f'W2.grad shape: {W2.grad.shape}')  # (2, 4)
print(f'Max |grad W1|: {W1.grad.abs().max().item():.4f}')
print(f'Max |grad W2|: {W2.grad.abs().max().item():.4f}')
```

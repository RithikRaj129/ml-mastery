**Example 2.8.1**

```python
from scipy import stats
import numpy as np

# Gaussian (Normal) distribution
mu, sigma = 0.0, 1.0
norm = stats.norm(mu, sigma)

print('=== Gaussian N(0,1) ===')
print(f'PDF at x=0:         {norm.pdf(0):.4f}')    # 0.3989
print(f'CDF at x=0:         {norm.cdf(0):.4f}')    # 0.5000
print(f'P(-1≤X≤1):          {norm.cdf(1)-norm.cdf(-1):.4f}')  # 0.6827 (68% rule)
print(f'P(-2≤X≤2):          {norm.cdf(2)-norm.cdf(-2):.4f}')  # 0.9545
print(f'P(-3≤X≤3):          {norm.cdf(3)-norm.cdf(-3):.4f}')  # 0.9973
print(f'95th percentile:    {norm.ppf(0.95):.4f}')  # 1.6449

# Bernoulli distribution
bern = stats.bernoulli(0.3)   # P(X=1) = 0.3, P(X=0) = 0.7
print('\n=== Bernoulli(p=0.3) ===')
print(f'P(X=1): {bern.pmf(1):.2f}  P(X=0): {bern.pmf(0):.2f}')
print(f'Mean:   {bern.mean():.2f}   Var: {bern.var():.4f}')   # E[X]=p, Var=p(1-p)

# Simulate sampling
samples = norm.rvs(size=10000, random_state=42)
print(f'\n10,000 N(0,1) samples: mean={samples.mean():.3f}, std={samples.std():.3f}')
```

**Example 2.8.2**

```python
np.random.seed(0)
X = np.random.normal(5, 2, 1000)  # N(μ=5, σ=2) - mean 5, std 2
Y = 0.7*X + np.random.normal(0, 1, 1000)  # Y is correlated with X

print(f'E[X]   = {X.mean():.3f}  (true: 5.000)')
print(f'Var(X) = {X.var():.3f}  (true: 4.000)')
print(f'Std(X) = {X.std():.3f}  (true: 2.000)')
print(f'Cov(X,Y) = {np.cov(X,Y)[0,1]:.3f}')
print(f'Corr(X,Y)= {np.corrcoef(X,Y)[0,1]:.3f}')

# Full covariance matrix for a dataset
data = np.column_stack([X, Y, np.random.normal(0,1,1000)])  # 3 features
cov_matrix = np.cov(data.T)
print(f'\nCovariance matrix (3×3):')
print(np.round(cov_matrix, 3))
```

**Example 2.9.1**

```python
# Conditional probability with a concrete example
# Medical test: disease prevalence 1%, test sensitivity 95%, specificity 90%

P_disease    = 0.01    # Prior: P(D)  - 1% of population has the disease
P_pos_given_D = 0.95   # Sensitivity: P(+|D) - 95% true positive rate
P_pos_given_nD= 0.10   # 1-Specificity: P(+|¬D) - 10% false positive rate
P_no_disease  = 1 - P_disease

# Total probability of a positive test
P_pos = P_pos_given_D * P_disease + P_pos_given_nD * P_no_disease
print(f'P(+ test) = {P_pos:.4f}  ({P_pos:.1%})')

# Conditional: P(Disease | + test) via Bayes
P_D_given_pos = (P_pos_given_D * P_disease) / P_pos
print(f'P(Disease|+test) = {P_D_given_pos:.4f}  ({P_D_given_pos:.1%})')

# Surprising result: even with a 95% sensitive test,
# a positive result only means ~8.7% chance of actually having disease
# This is the base rate fallacy - rare diseases make false positives dominate
print(f'\nIntuition: most positives are false positives because disease is rare.')
print(f'True positives:  {P_pos_given_D*P_disease*10000:.0f} per 10,000 people')
print(f'False positives: {P_pos_given_nD*P_no_disease*10000:.0f} per 10,000 people')
```

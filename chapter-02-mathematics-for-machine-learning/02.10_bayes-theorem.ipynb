```python
from sklearn.naive_bayes import MultinomialNB, GaussianNB
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score, classification_report
from sklearn.model_selection import train_test_split

# Text classification with Naive Bayes - 20 Newsgroups dataset
categories = ['sci.med', 'sci.space', 'comp.graphics', 'rec.sport.hockey']
data = fetch_20newsgroups(subset='all', categories=categories, random_state=42)

# TF-IDF vectorisation
vectorizer = TfidfVectorizer(max_features=5000, stop_words='english')
X_tfidf = vectorizer.fit_transform(data.data)
y = data.target
X_train, X_test, y_train, y_test = train_test_split(
X_tfidf, y, test_size=0.2, random_state=42, stratify=y)

# Naive Bayes: assumes feature independence given class
nb = MultinomialNB(alpha=0.1)   # alpha: Laplace smoothing
nb.fit(X_train, y_train)
y_pred = nb.predict(X_test)
print(f'Accuracy: {accuracy_score(y_test, y_pred):.4f}')
print(classification_report(y_test, y_pred, target_names=categories))
```

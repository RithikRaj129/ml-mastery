**Example 2.11.1**

```python
from scipy import stats
import numpy as np

data_sym  = np.array([2,3,3,4,4,4,5,5,6])           # Roughly symmetric
data_skew = np.array([1,1,2,2,2,3,3,5,50])           # Right-skewed (outlier)

for name, d in [('Symmetric', data_sym), ('Skewed', data_skew)]:
    print(f'\n{name} data: {d}')
    print(f'  Mean:   {d.mean():.2f}')
    print(f'  Median: {np.median(d):.2f}')
    print(f'  Mode:   {stats.mode(d, keepdims=True).mode[0]}')
    print(f'  Geometric mean: {stats.gmean(d):.2f}')
```

**Example 2.11.2**

```python
data = np.array([2,4,4,4,5,5,7,9,10,14,15,50])  # with outlier

Q1, Q2, Q3 = np.percentile(data, [25, 50, 75])
IQR = Q3 - Q1
print(f'Q1={Q1}  Q2(Median)={Q2}  Q3={Q3}  IQR={IQR}')
print(f'Min={data.min()}  Max={data.max()}')

# Outlier detection using Tukey fences
lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR
outliers = data[(data < lower_fence) | (data > upper_fence)]
print(f'Tukey fences: [{lower_fence:.1f}, {upper_fence:.1f}]')
print(f'Outliers detected: {outliers}')
```

**Example 2.11.3**

```python
data = np.array([2.,4.,4.,4.,5.,5.,7.,9.,10.,14.])

print(f'Mean:      {data.mean():.4f}')
print(f'Variance:  {data.var():.4f}   (biased, /n)')
print(f'Variance:  {data.var(ddof=1):.4f}  (unbiased, /n-1)')
print(f'Std Dev:   {data.std(ddof=1):.4f}')
print(f'MAD:       {np.median(np.abs(data - np.median(data))):.4f}')
print(f'Range:     {data.max()-data.min():.0f}')
print(f'CV:        {data.std()/data.mean()*100:.1f}%')

# StandardScaler: subtract mean, divide by std - makes each feature N(0,1)
data_scaled = (data - data.mean()) / data.std(ddof=1)
print(f'\nScaled mean: {data_scaled.mean():.4f}')
print(f'Scaled std:  {data_scaled.std(ddof=1):.4f}')
```

**Example 2.11.4**

```python
from scipy import stats

# Three distributions with different shapes
np.random.seed(0)
normal    = np.random.normal(0, 1, 1000)
right_skew= np.random.exponential(1, 1000)      # Right-skewed
heavy_tail= np.random.standard_t(df=3, size=1000) # Heavy tails (t with 3 dof)

for name, d in [('Normal', normal), ('Right-skewed', right_skew), ('Heavy-tail', heavy_tail)]:
    sk = stats.skew(d)
    ku = stats.kurtosis(d)  # Excess kurtosis (normal=0)
    jb_p = stats.jarque_bera(d).pvalue
    print(f'{name:<14}: skew={sk:+.3f}  excess_kurtosis={ku:+.3f}  JB p={jb_p:.4f}')
```

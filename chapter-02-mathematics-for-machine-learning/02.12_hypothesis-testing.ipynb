**Example 2.12.1**

```python
from scipy import stats
import numpy as np

# 95% CI for mean accuracy across 10 cross-validation folds
cv_accuracies = np.array([0.91, 0.93, 0.92, 0.90, 0.94,
0.92, 0.91, 0.93, 0.92, 0.91])

n    = len(cv_accuracies)
mean = cv_accuracies.mean()
se   = cv_accuracies.std(ddof=1) / np.sqrt(n)   # standard error

# t-interval (small sample, unknown population std)
t_crit = stats.t.ppf(0.975, df=n-1)             # Two-tailed 95% CI
ci_low  = mean - t_crit * se
ci_high = mean + t_crit * se

print(f'Mean accuracy: {mean:.4f} ({mean:.1%})')
print(f'Std deviation: {cv_accuracies.std(ddof=1):.4f}')
print(f'Standard error:{se:.4f}')
print(f'95% CI: [{ci_low:.4f}, {ci_high:.4f}]')
print(f'       = [{ci_low:.1%}, {ci_high:.1%}]')

# Using scipy for verification
ci = stats.t.interval(0.95, df=n-1, loc=mean, scale=se)
print(f'scipy CI: [{ci[0]:.4f}, {ci[1]:.4f}]')
```

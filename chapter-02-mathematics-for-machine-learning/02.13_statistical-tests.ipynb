**Example 2.13.1**

```python
from scipy import stats
import numpy as np

np.random.seed(42)
# Model A: 20 cross-val folds
model_A = np.array([0.912,0.924,0.918,0.931,0.908,0.921,0.915,0.927,
                    0.919,0.922,0.914,0.925,0.914,0.920,0.923,0.916,
                    0.911,0.928,0.913,0.926])
# Model B: 20 cross-val folds
model_B = np.array([0.905,0.911,0.908,0.919,0.903,0.914,0.907,0.916,
                    0.910,0.913,0.906,0.918,0.909,0.912,0.915,0.907,
                    0.904,0.914,0.906,0.915])

# Independent t-test
t_stat, p_val = stats.ttest_ind(model_A, model_B)
print(f'Model A mean: {model_A.mean():.4f} ± {model_A.std():.4f}')
print(f'Model B mean: {model_B.mean():.4f} ± {model_B.std():.4f}')
print(f'Difference:   {model_A.mean()-model_B.mean():+.4f}')
print(f't-statistic:  {t_stat:.4f}')
print(f'p-value:      {p_val:.6f}')
print(f'Significant (α=0.05)? {"Yes" if p_val < 0.05 else "No"}')

# Paired t-test (same folds - more powerful if models are tested on same data)
t_paired, p_paired = stats.ttest_rel(model_A, model_B)
print(f'\nPaired t-test p-value: {p_paired:.6f}')

# Cohen's d - effect size
pooled_std = np.sqrt((model_A.std()**2 + model_B.std()**2)/2)
cohens_d   = (model_A.mean() - model_B.mean()) / pooled_std
print(f'Cohen\'s d: {cohens_d:.3f}  (small<0.2, medium<0.5, large>0.8)')
```

**Example 2.13.2**

```python
# Example: Is prediction error related to income bracket?
# Observed: [correct predictions, incorrect predictions] for each income group
observed = np.array([[450, 50],     # Low income: 450 correct, 50 wrong
                     [380, 120],    # Mid income: 380 correct, 120 wrong
                     [340, 160]])   # High income: 340 correct, 160 wrong

chi2, p_val, dof, expected = stats.chi2_contingency(observed)
print('Observed:'); print(observed)
print(f'\nExpected (under independence):')
print(np.round(expected, 1))
print(f'\nChi² = {chi2:.4f}')
print(f'DOF  = {dof}')
print(f'p    = {p_val:.6f}')
print(f'Model accuracy differs significantly by income group: {p_val<0.05}')
```

**Example 2.13.3**

```python
# Compare two models' error distributions (non-normal - skewed errors)
np.random.seed(1)
model_A_errors = np.abs(np.random.exponential(0.5, 100))   # Skewed
model_B_errors = np.abs(np.random.exponential(0.7, 100))   # Larger errors

# Shapiro-Wilk normality test - confirm non-normality
_, p_norm_A = stats.shapiro(model_A_errors)
_, p_norm_B = stats.shapiro(model_B_errors)
print(f'Shapiro-Wilk: Model A p={p_norm_A:.4f}, Model B p={p_norm_B:.4f}')
print(f'Normal? A={p_norm_A>0.05}, B={p_norm_B>0.05}  → use non-parametric test')

# Mann-Whitney U test
u_stat, p_mw = stats.mannwhitneyu(model_A_errors, model_B_errors,
                                   alternative='less')  # A errors < B errors
print(f'\nMann-Whitney U = {u_stat:.0f}')
print(f'p-value = {p_mw:.6f}')
print(f'Model A significantly lower errors: {p_mw<0.05}')
```

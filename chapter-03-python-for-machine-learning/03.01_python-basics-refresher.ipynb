**Example 3.1.1**

```python
# Comments start with # and are ignored by the interpreter
# Python uses indentation (4 spaces) to define blocks - no braces needed

# Variable assignment - no type declarations
x = 10
name = 'Alice'
is_valid = True

# Multiple assignment on one line
a, b, c = 1, 2, 3
x = y = z = 0    # All three variables set to 0

# Print function - the basic output tool
print('Hello, Machine Learning!')      # Hello, Machine Learning!
print(f'x = {x}, name = {name}')      # f-strings: x = 10, name = Alice
print(f'a+b+c = {a+b+c}')             # Expressions inside f-strings: 6

# Python is case-sensitive
Age = 25
age = 30
print(Age, age)    # 25 30 - these are different variables

# The type() function returns the type of any object
print(type(x))          # <class 'int'>
print(type(name))       # <class 'str'>
print(type(is_valid))   # <class 'bool'>
print(type(3.14))       # <class 'float'>
```

```python
# Jupyter magic commands - special commands starting with % or %%

%matplotlib inline      # Show plots inline in the notebook
%timeit sum(range(1000)) # Time a single-line expression (runs multiple times)
%time my_function()     # Time a single call
%who                    # List all variables in namespace
%whos                   # List variables with details
%reset                  # Clear all variables

# %%timeit magic on first line times the entire cell
# %%writefile script.py saves the cell to a file

# Useful Jupyter tricks for ML
import pandas as pd
import numpy as np

# Display settings
pd.set_option('display.max_columns', 50)   # Show up to 50 columns
pd.set_option('display.float_format', '{:.4f}'.format)  # 4 decimal places
np.set_printoptions(precision=4, suppress=True)  # Clean NumPy output
```

**Example 3.2.1**

```python
# ── Integer ──────────────────────────────────────────────────────────
n_samples  = 1000
n_features = 30
epoch      = 0
print(type(n_samples))       # <class 'int'>
print(2**32)                 # Python integers have unlimited precision
print(0xFF)                  # Hex literal: 255
print(0b1010)                # Binary literal: 10
# ── Float ────────────────────────────────────────────────────────────
learning_rate = 0.001
accuracy      = 0.9234
inf_val       = float('inf')   # Infinity - appears in log(0) scenarios
nan_val       = float('nan')   # Not a Number - missing data sentinel
print(1/3)                     # 0.3333333333333333 (64-bit double)
print(round(1/3, 4))           # 0.3333
print(f'{0.1 + 0.2:.14f}')    # 0.30000000000000004 - float precision!
# ── String ───────────────────────────────────────────────────────────
model_name = 'RandomForest'
path       = r'C:\Users\data\model.pkl'  # raw string - no escape processing
multi_line = '''Line 1
Line 2
Line 3'''

# String methods most used in ML/data science
text = '  Hello, World!  '
print(text.strip())             # 'Hello, World!'  - remove whitespace
print(text.lower())             # '  hello, world!  '
print(text.upper())             # '  HELLO, WORLD!  '
print(text.replace(',', ';'))   # '  Hello; World!  '
print(text.split(','))          # ['  Hello', ' World!  ']
print(','.join(['a','b','c']))  # 'a,b,c'  - join list into string
print('forest' in model_name)  # True - substring check
print(len(model_name))         # 12

# f-string formatting - the modern way
loss   = 0.04523
epoch  = 42
print(f'Epoch {epoch:03d}: loss = {loss:.6f}')  # Epoch 042: loss = 0.045230
print(f'Accuracy: {0.9234:.1%}')                # Accuracy: 92.3%
print(f'Params:   {1_234_567:,}')               # Params:   1,234,567
# ── Boolean ──────────────────────────────────────────────────────────
is_training  = True
use_gpu      = False
print(int(True), int(False))   # 1 0 - bool is a subclass of int
print(True + True)             # 2 - booleans behave as 0/1 in arithmetic
# In ML: sum(y_pred == y_true) counts correct predictions
y_pred = [1,0,1,1,0]
y_true = [1,1,1,0,0]
correct = sum(p==t for p,t in zip(y_pred, y_true))
print(f'Accuracy: {correct}/{len(y_true)} = {correct/len(y_true):.0%}')  # 60%
```

**Example 3.2.2**

```python
# Creating lists
features    = ['age', 'income', 'tenure', 'clicks']
loss_history= [0.91, 0.74, 0.63, 0.54, 0.48, 0.43]
mixed       = [1, 'hello', True, 3.14, [1,2]]   # Any types, even nested

# Indexing - 0-based; negative indexing from the end
print(features[0])      # 'age'   - first element
print(features[-1])     # 'clicks' - last element
print(features[1:3])    # ['income', 'tenure'] - slice [start:stop)
print(features[::2])    # ['age', 'tenure'] - every 2nd element
print(features[::-1])   # Reversed list

# Mutating lists
loss_history.append(0.39)          # Add to end
loss_history.insert(0, 1.0)        # Insert at index 0
loss_history.extend([0.35, 0.32])  # Add multiple items
loss_history.pop()                 # Remove and return last item
loss_history.remove(1.0)           # Remove first occurrence of value

# Useful list operations
print(len(loss_history))           # Length
print(min(loss_history))           # Minimum loss
print(max(loss_history))           # Maximum loss
print(sorted(loss_history))        # Sorted copy (original unchanged)
loss_history.sort(reverse=True)    # Sort in place

# List comprehensions - concise and Pythonic
squares      = [x**2 for x in range(10)]
even_squares = [x**2 for x in range(10) if x % 2 == 0]
normalised   = [x/max(loss_history) for x in loss_history]
print(squares)       # [0,1,4,9,16,25,36,49,64,81]
print(even_squares)  # [0,4,16,36,64]

# Unpacking
first, *rest = features
print(first)   # 'age'
print(rest)    # ['income', 'tenure', 'clicks']
```

**Example 3.2.3**

```python
# Creating tuples
input_shape = (224, 224, 3)   # Image: height × width × channels
train_test  = (0.8, 0.2)      # Train/test split ratio
rgb         = (255, 128, 0)   # RGB colour

# Tuples behave like lists for reading
print(input_shape[0])         # 224
print(input_shape[-1])        # 3 (channels)
h, w, c = input_shape         # Unpacking
print(f'{h}×{w}×{c}')        # 224×224×3

# Tuples as dictionary keys (lists cannot be dict keys - they're not hashable)
coord_label = {(0,0): 'origin', (1,0): 'x-axis', (0,1): 'y-axis'}
print(coord_label[(0,0)])     # 'origin'

# Named tuples - like lightweight classes
from collections import namedtuple
ModelResult = namedtuple('ModelResult', ['accuracy','precision','recall','f1'])
result = ModelResult(accuracy=0.94, precision=0.92, recall=0.96, f1=0.94)
print(result.accuracy)        # 0.94  - access by name
print(result[0])              # 0.94  - access by index
print(result._asdict())       # OrderedDict - convert to dict
```

**Example 3.2.4**

```python
# Creating dictionaries
hyperparams = {
    'learning_rate': 0.001,
    'batch_size':    32,
    'n_epochs':      100,
    'dropout':       0.3,
    'optimizer':     'adam'
}

# Accessing values
print(hyperparams['learning_rate'])           # 0.001
print(hyperparams.get('weight_decay', 0.0))  # 0.0 - default if key missing

# Modifying
hyperparams['weight_decay'] = 1e-4           # Add new key
hyperparams['batch_size']   = 64             # Update existing key
del hyperparams['dropout']                   # Delete a key

# Iterating
for key, value in hyperparams.items():
    print(f'  {key:20s}: {value}')

print('\nKeys:  ', list(hyperparams.keys()))
print('Values:', list(hyperparams.values()))
print('learning_rate' in hyperparams)        # True - membership test

# Dict comprehensions
metrics = {'accuracy': 0.94, 'precision': 0.91, 'recall': 0.96, 'f1': 0.935}
pct_metrics = {k: f'{v:.1%}' for k, v in metrics.items()}
print(pct_metrics)

# Merging dicts (Python 3.9+)
base_config = {'lr': 0.001, 'epochs': 100}
override    = {'epochs': 50, 'batch': 32}
final_config = base_config | override   # override takes priority
print(final_config)

# defaultdict - avoids KeyError for missing keys
from collections import defaultdict
class_counts = defaultdict(int)            # Default value is int() = 0
labels = ['cat','dog','cat','bird','dog','cat']
for label in labels:
    class_counts[label] += 1
print(dict(class_counts))   # {'cat': 3, 'dog': 2, 'bird': 1}
```

**Example 3.2.5**

```python
# Creating sets
actual_labels    = {'cat','dog','bird','cat','dog','cat'}  # Duplicates removed
predicted_labels = {'cat','fish','bird','dog'}

print(actual_labels)               # {'cat', 'dog', 'bird'} - order not guaranteed
print(len(actual_labels))          # 3 - unique only

# Set operations - essential for class analysis
print(actual_labels & predicted_labels)  # Intersection: {'cat','dog','bird'}
print(actual_labels | predicted_labels)  # Union: {'cat','dog','bird','fish'}
print(actual_labels - predicted_labels)  # Difference: set() - all matched
print(predicted_labels - actual_labels)  # {'fish'} - predicted but never actual

# Check for unseen classes at inference time
train_classes = {'cat','dog','bird','fish'}
test_classes  = {'cat','dog','rabbit'}  # 'rabbit' was never in training!
unseen = test_classes - train_classes
if unseen:
    print(f'Warning: unseen classes at inference: {unseen}')

# Fast deduplication of feature lists
feature_list = ['age','income','age','clicks','income','tenure']
unique_features = list(set(feature_list))  # Remove duplicates
print(f'Removed {len(feature_list)-len(unique_features)} duplicates')
```

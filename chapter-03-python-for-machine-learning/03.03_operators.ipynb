**Example 3.3.1**

```python
a, b = 14, 5
print(a + b)    # 22   - Addition
print(a - b)    # 12   - Subtraction
print(a * b)    # 85   - Multiplication
print(a / b)    # 3.4  - True division (always float in Python 3)
print(a // b)   # 3    - Floor division (integer result)
print(a % b)    # 2    - Modulo (remainder)
print(a ** b)   # 1419857 - Exponentiation (14^5)

# Augmented assignment operators
loss = 1.0
loss -= 0.1    # loss = loss - 0.1 = 0.9
loss *= 0.95   # loss = loss * 0.95
loss += 0.01   # loss = loss + 0.01
print(f'loss = {loss:.4f}')   # 0.8650

# ML-relevant arithmetic patterns
n_samples = 10000
batch_size = 32
n_batches = n_samples // batch_size   # Integer division for batch count
remainder  = n_samples % batch_size   # Leftover samples in last batch
print(f'{n_batches} full batches, {remainder} remaining samples')

# Percentage error
y_true, y_pred = 100.0, 95.0
pct_error = abs(y_true - y_pred) / y_true * 100
print(f'MAPE: {pct_error:.1f}%')
```

**Example 3.3.2**

```python
# Comparison operators - return bool
accuracy = 0.94
threshold = 0.90

print(accuracy > threshold)    # True
print(accuracy == 0.94)        # True  (== for equality, not = which is assignment)
print(accuracy != 1.0)         # True
print(0.90 <= accuracy <= 1.0) # True  - Python supports chained comparisons!

# Logical operators: and, or, not
has_gpu       = True
memory_gb     = 16
model_loaded  = False

can_train = has_gpu and memory_gb >= 8 and not model_loaded
print(f'Can start training: {can_train}')  # True

# Short-circuit evaluation
# 'and' stops at first False; 'or' stops at first True
def expensive_check(): print('Running check...'); return True

# expensive_check() never called - short-circuits at False
result = False and expensive_check()
print(result)   # False (no 'Running check...' printed)

# Walrus operator := (Python 3.8+) - assign and test in one
import re
text = 'loss: 0.0423'
if match := re.search(r'loss: ([\d.]+)', text):
    print(f'Found loss value: {match.group(1)}')

# Identity operators: is, is not
a = None
if a is None:       # Correct way to check for None
    print('a is None')
if a is not None:   # Correct negation
    print('a has a value')

# Membership operators: in, not in
valid_optimizers = ['adam','sgd','rmsprop','adagrad']
optimizer = 'adam'
if optimizer in valid_optimizers:
    print(f'{optimizer} is a valid optimizer')
```

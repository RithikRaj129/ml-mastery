**Example 3.4.1**

```python
# Basic if / elif / else
def classify_loss(loss):
    if loss < 0.1:
        return 'Excellent - model is converging well'
    elif loss < 0.3:
        return 'Good - continue training'
    elif loss < 0.5:
        return 'Moderate - consider tuning hyperparameters'
    elif loss < 1.0:
        return 'High - check learning rate and data'
    else:
        return 'Critical - model may have diverged'

for loss_val in [0.05, 0.25, 0.45, 0.85, 2.5]:
    print(f'Loss {loss_val}: {classify_loss(loss_val)}')

# Ternary (conditional) expression - single-line if/else
accuracy = 0.93
status = 'PASS' if accuracy >= 0.90 else 'FAIL'
print(f'Model status: {status}')   # PASS

# Nested conditions - model selection logic
data_size = 100_000
n_features = 50
has_gpu = True

if data_size > 50_000:
    if has_gpu and n_features > 20:
        recommendation = 'Deep Learning (neural network)'
    else:
        recommendation = 'XGBoost or LightGBM'
else:
    if n_features < 10:
        recommendation = 'Logistic Regression'
    else:
        recommendation = 'Random Forest'

print(f'Recommended model: {recommendation}')
```

**Example 3.4.2**

```python
# Iterating over a list
metrics = ['accuracy', 'precision', 'recall', 'f1']
scores  = [0.94, 0.91, 0.96, 0.935]

# zip - iterate two lists in parallel
for metric, score in zip(metrics, scores):
    print(f'{metric:12s}: {score:.3f}')

# enumerate - get index and value simultaneously
print('\nEpoch training loop:')
loss_history = [0.91, 0.74, 0.63, 0.54]
for epoch, loss in enumerate(loss_history, start=1):
    delta = '' if epoch == 1 else f' (Δ{loss-loss_history[epoch-2]:+.2f})'
    print(f'  Epoch {epoch}: loss={loss:.2f}{delta}')

# range - generate integer sequences
print('\nRange examples:')
print(list(range(5)))          # [0,1,2,3,4]
print(list(range(2, 10, 2)))   # [2,4,6,8]  - start, stop, step
print(list(range(10, 0, -2)))  # [10,8,6,4,2]  - countdown

# Iterating over dict items
hyperparams = {'lr': 0.001, 'batch': 32, 'epochs': 100}
for param, value in hyperparams.items():
    print(f'  {param}: {value}')

# Comprehensions - the Pythonic loop
# List comprehension
squared   = [x**2 for x in range(10)]
# Dict comprehension
pct_dict  = {m: f'{s:.0%}' for m, s in zip(metrics, scores)}
# Set comprehension
unique_mods = {len(m) for m in metrics}  # Unique word lengths
# Generator expression - lazy evaluation (memory efficient for large data)
total_loss = sum(x**2 for x in loss_history)  # Doesn't build a list

print(pct_dict)
print(f'Sum of squared losses: {total_loss:.4f}')
```

**Example 3.4.3**

```python
# Training until convergence
import random; random.seed(42)

loss = 1.0
epoch = 0
target_loss = 0.5
max_epochs = 1000

while loss > target_loss and epoch < max_epochs:
    loss -= random.uniform(0.02, 0.08)  # Simulate training step
    loss = max(loss, 0.0)               # Can't go below 0
    epoch += 1
    if epoch % 5 == 0:
        print(f'Epoch {epoch:3d}: loss = {loss:.4f}')

print(f'\nConverged after {epoch} epochs. Final loss: {loss:.4f}')

# break and continue
print('\nSearching for best hyperparameter:')
for lr in [1.0, 0.1, 0.01, 0.001, 0.0001]:
    if lr > 0.1:
        print(f'  lr={lr}: skipping (too large)')
        continue             # Skip to next iteration
    val_accuracy = 0.90 + lr * 10   # Simulated
    print(f'  lr={lr}: val_acc={val_accuracy:.3f}')
    if val_accuracy > 0.92:
        print(f'  Found good lr={lr}!')
        break               # Exit loop early
```

**Example 3.11.1**

```python
import matplotlib.pyplot as plt
import numpy as np

# ── The Figure/Axes Architecture ─────────────────────────────────
# Figure: the entire canvas
# Axes: individual plot area (a Figure can have multiple Axes = subplots)

fig, axes = plt.subplots(2, 3, figsize=(15, 8))   # 2 rows × 3 cols
fig.suptitle('ML Visualization Gallery', fontsize=16, fontweight='bold', y=1.02)

np.random.seed(42)

# ── Plot 1: Learning Curves ───────────────────────────────────────
ax = axes[0, 0]
epochs = np.arange(1, 101)
train_loss = np.exp(-epochs/20) + 0.05 + 0.02*np.random.randn(100)
val_loss   = np.exp(-epochs/20) + 0.12 + 0.03*np.random.randn(100)
train_loss = np.clip(train_loss, 0.04, 1.0)
val_loss   = np.clip(val_loss, 0.10, 1.0)

ax.plot(epochs, train_loss, 'b-', linewidth=2, label='Train Loss', alpha=0.8)
ax.plot(epochs, val_loss,   'r--',linewidth=2, label='Val Loss',   alpha=0.8)
ax.fill_between(epochs, train_loss-0.01, train_loss+0.01, alpha=0.15, color='blue')
ax.axhline(y=0.1, color='green', linestyle=':', alpha=0.7, label='Target loss')
ax.set_xlabel('Epoch'); ax.set_ylabel('Loss')
ax.set_title('Training Curves'); ax.legend(); ax.grid(alpha=0.3)

# ── Plot 2: Distribution / Histogram ─────────────────────────────
ax = axes[0, 1]
class0 = np.random.normal(0, 1, 500)
class1 = np.random.normal(2, 1, 500)
ax.hist(class0, bins=30, alpha=0.6, color='blue',   label='Class 0', edgecolor='white')
ax.hist(class1, bins=30, alpha=0.6, color='orange', label='Class 1', edgecolor='white')
ax.axvline(x=1, color='red', linestyle='--', linewidth=2, label='Decision boundary')
ax.set_xlabel('Feature Value'); ax.set_ylabel('Count')
ax.set_title('Feature Distribution by Class'); ax.legend(); ax.grid(alpha=0.3)

# ── Plot 3: Scatter Plot ──────────────────────────────────────────
ax = axes[0, 2]
X_scatter = np.random.randn(200, 2)
labels    = (X_scatter[:,0] + X_scatter[:,1] > 0).astype(int)
colors    = ['steelblue' if l==0 else 'orangered' for l in labels]
sc = ax.scatter(X_scatter[:,0], X_scatter[:,1], c=colors, alpha=0.7,
               s=40, edgecolors='white', linewidth=0.5)
ax.set_xlabel('Feature 1'); ax.set_ylabel('Feature 2')
ax.set_title('2D Scatter - Binary Classification'); ax.grid(alpha=0.3)

# ── Plot 4: Bar Chart - Model Comparison ─────────────────────────
ax = axes[1, 0]
models  = ['LogReg','KNN','SVM','RF','XGB','NN']
acc     = [0.87, 0.85, 0.91, 0.93, 0.95, 0.96]
colors_bar = ['#2196F3' if a < 0.93 else '#4CAF50' for a in acc]
bars = ax.bar(models, acc, color=colors_bar, edgecolor='white', linewidth=0.5)
ax.set_ylim(0.80, 1.00); ax.set_ylabel('Test Accuracy')
ax.set_title('Model Comparison')
for bar, a in zip(bars, acc):  # Annotate bars
    ax.text(bar.get_x()+bar.get_width()/2, bar.get_height()+0.002,
            f'{a:.2f}', ha='center', va='bottom', fontsize=9, fontweight='bold')
ax.grid(axis='y', alpha=0.3)

# ── Plot 5: Confusion Matrix ──────────────────────────────────────
ax = axes[1, 1]
cm = np.array([[45, 5, 2],[3, 38, 4],[1, 6, 46]])  # 3-class confusion matrix
im = ax.imshow(cm, interpolation='nearest', cmap='Blues')
plt.colorbar(im, ax=ax)
classes = ['Cat','Dog','Bird']
ax.set_xticks(range(3)); ax.set_xticklabels(classes)
ax.set_yticks(range(3)); ax.set_yticklabels(classes)
ax.set_xlabel('Predicted'); ax.set_ylabel('Actual')
ax.set_title('Confusion Matrix')
thresh = cm.max() / 2
for i in range(3):
    for j in range(3):
        ax.text(j, i, str(cm[i,j]), ha='center', va='center',
               color='white' if cm[i,j]>thresh else 'black', fontweight='bold')

# ── Plot 6: Feature Importance ───────────────────────────────────
ax = axes[1, 2]
features   = ['concave pts','perimeter','radius','area','texture',
              'smoothness','symmetry','fractal dim']
importance  = sorted([0.15,0.11,0.10,0.09,0.07,0.06,0.05,0.04])[::-1]
colors_imp  = ['#E53935' if i > 0.09 else '#1976D2' for i in importance]
ax.barh(features, importance, color=colors_imp, edgecolor='white')
ax.set_xlabel('Importance Score'); ax.set_title('Feature Importance (Random Forest)')
ax.grid(axis='x', alpha=0.3)
ax.invert_yaxis()  # Highest importance at top

plt.tight_layout()
plt.savefig('ml_visualization_gallery.png', dpi=150, bbox_inches='tight')
plt.show()
print('Gallery saved: ml_visualization_gallery.png')
```

**Example 3.11.2**

```python
import matplotlib.pyplot as plt
import numpy as np

# Matplotlib style sheets - change overall appearance
# Available: 'seaborn-v0_8', 'ggplot', 'bmh', 'fivethirtyeight', 'dark_background'
plt.style.use('seaborn-v0_8-whitegrid')

# Professional ROC curve plot - common in ML presentations
np.random.seed(42)
n = 1000
y_true = np.random.choice([0,1], n, p=[0.7,0.3])

# Simulate predictions from 3 models
def simulate_scores(skill): return np.clip(y_true * skill + np.random.beta(1,3,n), 0, 1)
scores_a = simulate_scores(0.90)   # Good model
scores_b = simulate_scores(0.70)   # Average model
scores_c = np.random.rand(n)        # Random (AUC≈0.5)

from sklearn.metrics import roc_curve, auc

fig, ax = plt.subplots(figsize=(8, 7))

for name, scores, color, ls in [
    ('Model A (ResNet)',  scores_a, '#1976D2', '-'),
    ('Model B (XGBoost)', scores_b, '#388E3C', '--'),
    ('Random Baseline',   scores_c, '#E53935', ':'),
]:
    fpr, tpr, _ = roc_curve(y_true, scores)
    roc_auc = auc(fpr, tpr)
    ax.plot(fpr, tpr, color=color, ls=ls, lw=2.5,
           label=f'{name} (AUC = {roc_auc:.3f})')

ax.plot([0,1],[0,1], 'k--', lw=1, alpha=0.5)  # Diagonal
ax.fill_between([0,1],[0,1], alpha=0.05, color='gray')

ax.set_xlim([0.0, 1.0]); ax.set_ylim([0.0, 1.02])
ax.set_xlabel('False Positive Rate (1 - Specificity)', fontsize=13)
ax.set_ylabel('True Positive Rate (Sensitivity)', fontsize=13)
ax.set_title('Receiver Operating Characteristic Curve', fontsize=14, fontweight='bold')
ax.legend(loc='lower right', fontsize=11)
ax.grid(True, alpha=0.3)

# Add text annotation
ax.annotate('Perfect classifier\n(AUC = 1.0)', xy=(0,1), xytext=(0.2, 0.85),
           arrowprops=dict(arrowstyle='->', color='gray'),
           fontsize=10, ha='center', color='gray')

plt.tight_layout()
plt.savefig('roc_curve.png', dpi=150, bbox_inches='tight')
plt.show()
print('ROC curve saved.')
```

**Example 3.11.3**

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Load the built-in Titanic dataset
titanic = sns.load_dataset('titanic')
print(titanic.shape, titanic.columns.tolist())

sns.set_theme(style='whitegrid', palette='husl', font_scale=1.1)

fig, axes = plt.subplots(2, 3, figsize=(16, 10))
fig.suptitle('Seaborn Statistical Visualization - Titanic', fontsize=16, fontweight='bold')

# ── 1. Distribution plot ──────────────────────────────────────────
ax = axes[0,0]
sns.histplot(data=titanic, x='age', hue='survived', bins=30,
             kde=True, element='step', ax=ax, palette=['#E53935','#1976D2'])
ax.set_title('Age Distribution by Survival')

# ── 2. Box plot ───────────────────────────────────────────────────
ax = axes[0,1]
sns.boxplot(data=titanic, x='pclass', y='age', hue='survived',
           ax=ax, palette='Set2', width=0.6)
ax.set_title('Age by Class and Survival')
ax.set_xlabel('Passenger Class')

# ── 3. Violin plot ────────────────────────────────────────────────
ax = axes[0,2]
sns.violinplot(data=titanic.dropna(subset=['age']),
              x='pclass', y='age', hue='sex', split=True,
              ax=ax, palette='muted', inner='quartile')
ax.set_title('Age Distribution by Class and Sex')

# ── 4. Count plot ─────────────────────────────────────────────────
ax = axes[1,0]
sns.countplot(data=titanic, x='pclass', hue='survived',
             ax=ax, palette=['#EF5350','#42A5F5'], edgecolor='white')
ax.set_title('Passenger Count by Class')
ax.set_xlabel('Passenger Class'); ax.legend(title='Survived')

# ── 5. Heatmap - Correlation Matrix ──────────────────────────────
ax = axes[1,1]
corr_cols = ['survived','pclass','age','sibsp','parch','fare']
corr_matrix = titanic[corr_cols].corr()
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))  # Upper triangle
sns.heatmap(corr_matrix, ax=ax, mask=mask, annot=True, fmt='.2f',
           cmap='RdYlGn', center=0, linewidths=0.5,
           cbar_kws={'shrink':0.8})
ax.set_title('Feature Correlation Matrix')

# ── 6. Pair plot (separate figure - too large for subplot) ────────
ax = axes[1,2]
# Bar plot: survival rate by embarked port and class
survival = titanic.groupby(['embarked','pclass'])['survived'].mean().reset_index()
sns.barplot(data=survival, x='embarked', y='survived', hue='pclass',
           ax=ax, palette='Blues')
ax.set_title('Survival Rate by Port and Class')
ax.set_xlabel('Port of Embarkation'); ax.set_ylabel('Survival Rate')
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda y,_: f'{y:.0%}'))

plt.tight_layout()
plt.savefig('seaborn_gallery.png', dpi=150, bbox_inches='tight')
plt.show()
print('Seaborn gallery saved.')
```

**Example 3.11.4**

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Load breast cancer dataset
data = load_breast_cancer()
X = StandardScaler().fit_transform(data.data)
y = data.target

# ── PCA 2D scatter with class overlay ────────────────────────────
pca = PCA(n_components=2)
X_2d = pca.fit_transform(X)

pca_df = pd.DataFrame(X_2d, columns=['PC1','PC2'])
pca_df['Class'] = ['Benign' if l==1 else 'Malignant' for l in y]

fig, axes = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle('ML Analysis Plots - Breast Cancer Dataset', fontsize=14, fontweight='bold')

# PCA scatter
ax = axes[0]
sns.scatterplot(data=pca_df, x='PC1', y='PC2', hue='Class',
               palette={'Malignant':'#E53935','Benign':'#1976D2'},
               alpha=0.75, s=60, ax=ax)
ax.set_title(f'PCA Projection (Var explained: PC1={pca.explained_variance_ratio_[0]:.1%}, PC2={pca.explained_variance_ratio_[1]:.1%})')
ax.set_xlabel('First Principal Component')
ax.set_ylabel('Second Principal Component')

# ── Feature importance via correlation ───────────────────────────
ax = axes[1]
feature_df = pd.DataFrame(data.data, columns=data.feature_names)
feature_df['target'] = y

# Absolute correlation with target - proxy for feature importance
corr_with_target = feature_df.corr()['target'].drop('target').abs().sort_values(ascending=True)
top10 = corr_with_target.tail(10)

colors = sns.color_palette('RdYlGn_r', n_colors=len(top10))
sns.barplot(x=top10.values, y=top10.index, palette=colors[::-1], ax=ax)
ax.set_xlabel('|Correlation with Target|')
ax.set_title('Top 10 Features by Correlation with Target')
ax.grid(axis='x', alpha=0.3)

# Annotate bars
for i, v in enumerate(top10.values):
    ax.text(v + 0.01, i, f'{v:.3f}', va='center', fontsize=9)

plt.tight_layout()
plt.savefig('ml_analysis_plots.png', dpi=150, bbox_inches='tight')
plt.show()
print('ML analysis plots saved.')

# ── Pair plot for multivariate exploration (standalone) ───────────
feature_subset = pd.DataFrame(
    data.data[:, :4],
    columns=data.feature_names[:4]
)
feature_subset['diagnosis'] = ['Benign' if l==1 else 'Malignant' for l in y]

g = sns.pairplot(
    feature_subset,
    hue='diagnosis',
    palette={'Malignant':'#E53935','Benign':'#1976D2'},
    diag_kind='kde',
    plot_kws={'alpha': 0.6, 's': 30},
    corner=True    # Only lower triangle - avoids redundancy
)
g.fig.suptitle('Pair Plot - Top 4 Features', y=1.02, fontsize=13)
g.savefig('pairplot.png', dpi=120, bbox_inches='tight')
plt.show()
print('Pair plot saved.')
```

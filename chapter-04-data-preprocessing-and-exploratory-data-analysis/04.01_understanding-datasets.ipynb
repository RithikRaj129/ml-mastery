```python
import pandas as pd
import numpy as np

# ── CSV - most common format ──────────────────────────────────────
df = pd.read_csv('data.csv',
    sep=',',             # Delimiter (could be tab '\t' or semicolon ';')
    header=0,            # Row number for column names (None if no header)
    index_col=None,      # Column to use as row index
    dtype={'age': int, 'income': float},   # Explicit column types
    na_values=['NA','N/A','?','--',''],    # Strings to treat as NaN
    parse_dates=['signup_date'],            # Parse date columns
    nrows=1000,          # Load only first 1000 rows (for large files)
    usecols=['age','income','target']       # Load only specific columns
)

# ── Excel ─────────────────────────────────────────────────────────
df_excel = pd.read_excel('data.xlsx', sheet_name='Sheet1')

# ── JSON ──────────────────────────────────────────────────────────
df_json = pd.read_json('data.json', orient='records')

# ── SQL Database ──────────────────────────────────────────────────
# import sqlalchemy
# engine = sqlalchemy.create_engine('postgresql://user:pass@host/db')
# df_sql = pd.read_sql('SELECT * FROM customers WHERE active=1', engine)

# ── Parquet - columnar format for large datasets ──────────────────
# df_parquet = pd.read_parquet('data.parquet')  # pip install pyarrow

# ── First look at any dataset - always run these immediately ──────
def dataset_overview(df, name='Dataset'):
    print(f'{'='*50}')
    print(f'{name} Overview')
    print(f'{'='*50}')
    print(f'Shape:           {df.shape[0]:,} rows × {df.shape[1]} columns')
    print(f'Memory usage:    {df.memory_usage(deep=True).sum()/1024**2:.2f} MB')
    print(f'Duplicate rows:  {df.duplicated().sum():,}')
    print(f'Missing values:  {df.isnull().sum().sum():,} total')
    print(f'\nColumn overview:')
    overview = pd.DataFrame({
        'dtype':   df.dtypes,
        'non_null':df.count(),
        'missing': df.isnull().sum(),
        'miss_%':  (df.isnull().mean()*100).round(1),
        'unique':  df.nunique(),
        'sample':  [str(df[c].dropna().iloc[0]) if df[c].count()>0 else 'ALL NULL'
                    for c in df.columns]
    })
    print(overview.to_string())
    return overview

# Using the breast cancer dataset as example
from sklearn.datasets import load_breast_cancer
data = load_breast_cancer()
df_bc = pd.DataFrame(data.data, columns=data.feature_names)
df_bc['target'] = data.target
dataset_overview(df_bc, 'Breast Cancer')
```

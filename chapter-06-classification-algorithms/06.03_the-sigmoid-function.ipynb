**Example 6.3.1**

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(z):       return 1 / (1 + np.exp(-np.clip(z, -500, 500)))
def sigmoid_deriv(z): s = sigmoid(z); return s * (1 - s)

z = np.linspace(-8, 8, 500)
sig   = sigmoid(z)
dsig  = sigmoid_deriv(z)

fig, axes = plt.subplots(1, 3, figsize=(16, 5))
fig.suptitle('Sigmoid Function - Properties and Characteristics', fontsize=13, fontweight='bold')

# Plot 1: Sigmoid curve
ax = axes[0]
ax.plot(z, sig, 'b-', lw=3, label='σ(z)')
ax.axhline(0.5, color='red', ls='--', alpha=0.7, label='σ(0)=0.5 (decision threshold)')
ax.axhline(1.0, color='gray', ls=':', alpha=0.5, label='Upper bound: 1')
ax.axhline(0.0, color='gray', ls=':', alpha=0.5, label='Lower bound: 0')
ax.axvline(0, color='green', ls='--', alpha=0.5)
for z_val in [-4, -2, 0, 2, 4]:
    ax.annotate(f'σ({z_val})={sigmoid(z_val):.2f}', (z_val, sigmoid(z_val)),
               xytext=(z_val+0.3, sigmoid(z_val)-0.08), fontsize=8)
ax.set_xlabel('z (linear combination)'); ax.set_ylabel('σ(z) = P(y=1)')
ax.set_title('Sigmoid Function'); ax.legend(fontsize=9); ax.grid(alpha=0.3)

# Plot 2: Derivative
ax = axes[1]
ax.plot(z, dsig, 'r-', lw=3, label="σ'(z) = σ(z)(1-σ(z))")
ax.axvline(0, color='green', ls='--', alpha=0.5, label='Max at z=0: σ\'(0)=0.25')
ax.fill_between(z, dsig, alpha=0.15, color='red')
ax.set_xlabel('z'); ax.set_ylabel("σ'(z)")
ax.set_title('Sigmoid Derivative\n(Backpropagation gradient)')
ax.legend(fontsize=9); ax.grid(alpha=0.3)

# Plot 3: Log-odds (logit) relationship
ax = axes[2]
p = np.linspace(0.01, 0.99, 500)
logit = np.log(p / (1 - p))
ax.plot(p, logit, 'g-', lw=3, label='logit(p) = log(p/(1-p))')
ax.axhline(0, color='black', lw=1)
ax.axvline(0.5, color='red', ls='--', alpha=0.7, label='p=0.5 → logit=0')
ax.set_xlabel('Probability p'); ax.set_ylabel('Log-odds (logit)')
ax.set_title('Logit Function\n(Inverse of sigmoid)')
ax.legend(fontsize=9); ax.grid(alpha=0.3)

plt.tight_layout()
plt.savefig('sigmoid_function.png', dpi=150, bbox_inches='tight')
plt.show()

# Key values table
print(f'{'z':>6} {'σ(z)':>8} {'1-σ(z)':>10} {"σ'(z)":>8} Interpretation')
print('-'*60)
for z_val in [-4, -2, -1, 0, 1, 2, 4]:
    s = sigmoid(z_val); ds = sigmoid_deriv(z_val)
    interp = 'P(y=1)=50% - decision boundary' if z_val==0 else (
             'Very likely class 1' if z_val>2 else ('Very likely class 0' if z_val<-2 else 'Uncertain'))
    print(f'{z_val:>6} {s:>8.4f} {1-s:>10.4f} {ds:>8.4f}  {interp}')
```

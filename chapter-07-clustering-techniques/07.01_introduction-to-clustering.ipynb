```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.spatial.distance import cdist
from sklearn.preprocessing import StandardScaler

# Demonstrate distance metrics on simple 2D points
A = np.array([0.0, 0.0])
B = np.array([3.0, 4.0])

euclidean  = np.sqrt(np.sum((A-B)**2))
manhattan  = np.sum(np.abs(A-B))
cosine_sim = np.dot(A,B)/(np.linalg.norm(A)*np.linalg.norm(B)+1e-10)
cosine_dist= 1 - cosine_sim

print('Distance metrics between A=(0,0) and B=(3,4):')
print(f'  Euclidean:  {euclidean:.4f}  (geometric distance)')
print(f'  Manhattan:  {manhattan:.4f}  (sum of absolute differences)')
print(f'  Cosine Sim: {cosine_sim:.4f}  (angle between vectors)')

# Pairwise distance matrix for a small dataset
np.random.seed(42)
X = np.array([[1,2],[3,4],[5,1],[7,8],[2,6]])

print('\nPairwise Euclidean distance matrix:')
D = cdist(X, X, metric='euclidean')
print(np.round(D, 2))

print('\nPairwise Cosine distance matrix:')
D_cos = cdist(X, X, metric='cosine')
print(np.round(D_cos, 4))

# Impact of scaling on distances
print('\n=== Why Feature Scaling Matters for Distance ===')
X_unscaled = np.array([[25, 50000], [32, 80000], [28, 60000]])
X_scaled   = StandardScaler().fit_transform(X_unscaled)
print(f'Unscaled distance (p1,p2): {np.linalg.norm(X_unscaled[0]-X_unscaled[1]):.0f}')
print(f'Scaled distance   (p1,p2): {np.linalg.norm(X_scaled[0]-X_scaled[1]):.4f}')
print('Without scaling, income (range 50k-80k) dominates age (range 25-32) completely!')
```

**Example 8.1.1**

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.spatial.distance import cdist

# ── Demonstrate distance concentration ───────────────────────────
np.random.seed(42)
n_points = 200
dimensions = [2, 5, 10, 20, 50, 100, 200, 500, 1000]

print(f'{'Dim':>6} {'Mean Dist':>12} {'Max Dist':>12} {'Min Dist':>12} {'Contrast':>12}')
print('='*58)

means, contrasts = [], []
for d in dimensions:
    X = np.random.randn(n_points, d)
    D = cdist(X, X, metric='euclidean')
    np.fill_diagonal(D, np.inf)

    mean_d = np.mean(D[D < np.inf])
    max_d  = np.max(D[D < np.inf])
    min_d  = np.min(D[D < np.inf])
    contrast = (max_d - min_d) / min_d

    means.append(mean_d)
    contrasts.append(contrast)
    print(f'{d:>6} {mean_d:>12.4f} {max_d:>12.4f} {min_d:>12.4f} {contrast:>12.4f}')

# ── Data required grows exponentially ────────────────────────────
print('\nData needed to cover 10% of space (s=0.1^(1/d)):')
print(f'{'Dim':>6} {'Edge length':>14} {'Points for 1% density':>22}')
print('='*46)
for d in [1, 2, 3, 5, 10, 20, 50, 100]:
    s = 0.1 ** (1/d)
    # If unit hypercube has 100 uniform points in 1D, need 100^d in d-D
    points_needed = 100**d
    print(f'{d:>6} {s:>14.4f} {points_needed:>22,.0f}')

# ── Visualise distance concentration ─────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle('Curse of Dimensionality', fontsize=14, fontweight='bold')

ax = axes[0]
ax.plot(dimensions, contrasts, 'b-o', ms=8, lw=2.5)
ax.set_xlabel('Number of Dimensions (d)', fontsize=12)
ax.set_ylabel('Distance Contrast  (max-min)/min', fontsize=12)
ax.set_title('Distance Concentration\n(contrast → 0 as d increases)', fontsize=12)
ax.set_xscale('log'); ax.grid(alpha=0.3)
ax.axhline(0.5, color='red', ls='--', alpha=0.6, label='Contrast = 0.5 (threshold)')
ax.legend()

ax = axes[1]
d_range = np.arange(1, 21)
s_vals  = [0.1**(1/d) for d in d_range]
ax.plot(d_range, s_vals, 'r-o', ms=8, lw=2.5)
ax.fill_between(d_range, s_vals, 1.0, alpha=0.15, color='red', label='Uncovered space')
ax.set_xlabel('Number of Dimensions (d)', fontsize=12)
ax.set_ylabel('Edge length to cover 10% of data', fontsize=12)
ax.set_title('Sparsity Growth\n(need almost entire space to see 10% of data)', fontsize=12)
ax.grid(alpha=0.3); ax.legend()

plt.tight_layout()
plt.savefig('curse_dimensionality.png', dpi=150, bbox_inches='tight')
plt.show()
```

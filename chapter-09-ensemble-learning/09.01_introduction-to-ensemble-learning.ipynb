**Example 9.1.1**

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import comb

# ── Demonstrate ensemble error reduction ──────────────────────────
def ensemble_error(n_models, individual_error):
    '''P(majority vote wrong) for n independent classifiers with error rate ε.'''
    majority = int(n_models / 2) + 1
    error = 0
    for k in range(majority, n_models + 1):
        error += comb(n_models, k, exact=True) * (
            individual_error**k * (1-individual_error)**(n_models-k)
        )
    return error

print('Ensemble Error vs Number of Models')
print(f'Individual error: 0.35  (each model is 65% accurate)')
print(f'{'n_models':>10} {'Individual':>12} {'Ensemble':>12} {'Improvement':>14}')
print('='*52)

for n in [1, 3, 5, 11, 21, 51, 101, 201, 501]:
    ind = 0.35
    ens = ensemble_error(n, ind)
    imp = (ind - ens) / ind * 100
    print(f'{n:>10} {ind:>12.4f} {ens:>12.4f} {imp:>13.1f}%')

# ── Error vs individual model accuracy ────────────────────────────
print('\nEnsemble error (n=25) for different individual accuracy levels:')
print(f'{'Indiv Acc':>12} {'Indiv Error':>13} {'Ensemble Error':>15} {'Ensemble Acc':>14}')
print('='*58)
for acc in [0.50, 0.55, 0.60, 0.65, 0.70, 0.75, 0.80, 0.85, 0.90]:
    err = 1 - acc
    ens_err = ensemble_error(25, err)
    print(f'{acc:>12.0%} {err:>13.4f} {ens_err:>15.6f} {1-ens_err:>13.1%}')

# ── Visualise ensemble error reduction ────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle('Why Ensemble Learning Works', fontsize=14, fontweight='bold')

ax = axes[0]
n_range = range(1, 101, 2)  # Odd numbers only
for ind_err, color in [(0.25,'#43A047'),(0.35,'#1976D2'),(0.45,'#E53935')]:
    ens_errs = [ensemble_error(n, ind_err) for n in n_range]
    ax.plot(n_range, ens_errs, '-', lw=2.5, color=color,
           label=f'Individual error={ind_err:.2f}')
    ax.axhline(ind_err, color=color, ls='--', alpha=0.4)

ax.set_xlabel('Number of Models in Ensemble'); ax.set_ylabel('Ensemble Error Rate')
ax.set_title('Ensemble Error vs Number of Models\n(dashed = individual model error)')
ax.legend(); ax.grid(alpha=0.3)

ax = axes[1]
individual_errors = np.linspace(0.01, 0.49, 100)
for n, color in [(5,'#E53935'),(11,'#FF6F00'),(25,'#1976D2'),(101,'#43A047')]:
    ens_errs = [ensemble_error(n, e) for e in individual_errors]
    ax.plot(individual_errors, ens_errs, '-', lw=2.5, color=color, label=f'n={n}')
ax.plot([0,0.5],[0,0.5], 'k--', lw=1.5, alpha=0.6, label='No improvement line')
ax.fill_between([0,0.5],[0,0.5],[0,0], alpha=0.05, color='green', label='Improvement zone')
ax.set_xlabel('Individual Model Error'); ax.set_ylabel('Ensemble Error')
ax.set_title('Ensemble Error for Different Ensemble Sizes\n(below dashed = improvement)')
ax.legend(); ax.grid(alpha=0.3)

plt.tight_layout()
plt.savefig('ensemble_theory.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Architecture decision framework - Python pseudocode

def choose_serving_pattern(requirements):
    '''
    Choose between batch, real-time, and streaming inference.
    '''
    latency_ms   = requirements['max_latency_ms']
    throughput   = requirements['predictions_per_second']
    freshness    = requirements['feature_freshness_seconds']
    data_volume  = requirements['daily_predictions']

    # Decision tree
    if latency_ms > 60_000:  # > 1 minute acceptable
        return 'BATCH_SCORING'

    elif latency_ms > 5_000:  # 1-60 seconds
        if freshness < 3600:   # Features must be fresh (< 1 hour)
            return 'NEAR_REAL_TIME_STREAMING'
        else:
            return 'BATCH_SCORING'

    elif latency_ms > 100:  # 100ms - 5s
        if throughput > 10_000:  # High throughput
            return 'STREAMING_INFERENCE'
        else:
            return 'REAL_TIME_API'

    else:  # < 100ms (strict real-time)
        if throughput > 100_000:
            return 'STREAMING_INFERENCE_WITH_CACHING'
        else:
            return 'REAL_TIME_API_WITH_CACHING'


# Example use cases
use_cases = {
    'Email churn model': {
        'max_latency_ms': 86_400_000,  # 24 hours
        'predictions_per_second': 1,
        'feature_freshness_seconds': 86400,
        'daily_predictions': 1_000_000
    },
    'Fraud detection': {
        'max_latency_ms': 200,
        'predictions_per_second': 5000,
        'feature_freshness_seconds': 60,
        'daily_predictions': 432_000_000
    },
    'Product recommendation': {
        'max_latency_ms': 100,
        'predictions_per_second': 50_000,
        'feature_freshness_seconds': 300,
        'daily_predictions': 4_320_000_000
    },
    'Demand forecasting': {
        'max_latency_ms': 3_600_000,  # 1 hour
        'predictions_per_second': 0.1,
        'feature_freshness_seconds': 3600,
        'daily_predictions': 10_000
    },
}

print('Serving Pattern Recommendations:')
for case, reqs in use_cases.items():
    pattern = choose_serving_pattern(reqs)
    print(f'  {case:<28}: {pattern}')
```
